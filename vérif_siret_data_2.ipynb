# Pour rajouter une colonne qui indique si le siret est vrai ou pas

```python
import numpy as np
import pandas as pd
```

On veut reporter ces résultats dans data_set_2 et donc on extrait le numéro d'établissement et la colonne siret_vrai de dataset_1 (=data_verif_siret) et merge avec dataset_2

```python
#extraction des colonnes de dataset_1 qui est devenu 
dataset_1=pd.read_csv(r"resultats\data_vérif_siret.csv",sep=',',na_values=['[ND]'],low_memory=False)
data_1=dataset_1[['siret_vrai','num_etablissement']]
```

```python
#merge avec le dataset_2
data_2=pd.read_csv(r"resultats\dataset_final_2.csv",sep=',',na_values=['[ND]'],low_memory=False)
result_tableau=pd.merge(data_1,data_2,left_on='num_etablissement',right_on='num_etablissement')
```

```python
#enregistrement du fichier final
result_tableau.to_csv('resultats/data_vérif_siret_data_2.csv', index=False)
```
